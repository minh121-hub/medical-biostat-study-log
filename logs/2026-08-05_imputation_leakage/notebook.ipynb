```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split

df = pd.DataFrame({
    'age': [25, np.nan, 30, 45, np.nan, 22],
    'income': [3000, 3500, np.nan, 8000, 2900, np.nan],
    'gender': ['M', 'F', np.nan, 'M', 'F', 'F']
})

# 수치형: median으로 대치
num_imputer = SimpleImputer(strategy='median')

# 범주형: most_frequent로 대치
cat_imputer = SimpleImputer(strategy='most_frequent')

# 1. 먼저 train/test 분리
train_df, test_df = train_test_split(df, test_size=0.25, random_state=42)

# 2. imputer는 train으로만 fit
num_imputer = SimpleImputer(strategy='median')
train_df[['age', 'income']] = num_imputer.fit_transform(train_df[['age', 'income']])
# 이 순간 num_imputer 내에 train_set의 계산값이 저장됨

# 3. test set엔 transform만 (train에서 계산된 median 값을 그대로 적용)
test_df[['age', 'income']] = num_imputer.transform(test_df[['age', 'income']])

# test set의 결측치에 대치하는 값은 반드시 train으로만 계산해야 함.(test set은 모른다고 가정)
```
